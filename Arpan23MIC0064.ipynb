# Task 1


```python
import threading
import time
import concurrent.futures

start = time.perf_counter()

def do_something(sec):
    print(f'Sleeping for {sec} seconds')
    time.sleep(sec)
    print('Done Sleeping')
    return ""

t1 = threading.Thread(target = do_something, args = [1])
t1.start()
t1.join()


finish = time.perf_counter()
print(f'Finished in {round(finish-start,2)} seconds')
```

# Task 2


```python
start1 = time.perf_counter()
thread_list =[]

for _ in range(50):
    t2 = threading.Thread(target = do_something, args = [1])
    t2.start()
    thread_list.append(t2)
for i in thread_list:
    i.join()
finish1 = time.perf_counter()
print(f'Finished in {round(finish1-start1)} second')
```

# Task 4


```python
start = time.perf_counter()
with concurrent.futures.ThreadPoolExecutor() as executor:
    results = [executor.submit(do_something, 1) for _ in range(2)]
    for f in concurrent.futures.as_completed(results):
        print(f.result())

finish = time.perf_counter()
print(f'Finished in {round(finish-start, 2)} seconds')
```

# Task 7

```python
import concurrent.futures
import time
import urllib.request

audio_tasks = {
    "track_1.mp3": "https://samplelib.com/wav/sample-3s-8khz.wav",
    "track_2.mp3": "https://samplelib.com/wav/sample-3s-16khz.wav",
    "track_3.mp3": "https://samplelib.com/wav/sample-3s-stereo.wav"
}

def download_audio(filename, url):
    print(f"[START] Downloading {filename}...")
    try:
        
        urllib.request.urlretrieve(url, filename)
        return f"[SUCCESS] Finished downloading {filename}"
    except Exception as e:
        return f"[ERROR] Failed to download {filename}: {e}"


start_time = time.perf_counter()

with concurrent.futures.ThreadPoolExecutor() as executor:
    futures = [
        executor.submit(download_audio, filename, url) 
        for filename, url in audio_tasks.items()
    ]
    
    for future in concurrent.futures.as_completed(futures):
        print(future.result())

end_time = time.perf_counter()
print(f"\nAll downloads completed in {round(end_time - start_time, 2)} seconds.")
```
